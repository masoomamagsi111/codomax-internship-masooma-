```python

print("==============================")
print("Codomax Internship - Module 2")
print("Name: Masooma")
print("==============================")

import json, os
FILE = "/content/students.json"

def load_data():
    try:
        if not os.path.exists(FILE):
            with open(FILE, "w") as f:
                json.dump([], f)
            return []
        with open(FILE, "r") as f:
            return json.load(f)
    except:
        return []

def save_data(data):
    with open(FILE, "w") as f:
        json.dump(data, f, indent=4)

def add_student(roll, name, marks):
    data = load_data()
    if roll in [s['roll'] for s in data]:
        return "Already exists!"
    data.append({"roll": roll, "name": name, "marks": marks})
    save_data(data)
    return f"Student {name} Added!"

def view_students():
    return load_data()

def update_student(roll, new_name):
    data = load_data()
    for s in data:
        if s["roll"] == roll:
            s["name"] = new_name
            save_data(data)
            return "Updated!"
    return "Not Found!"

def delete_student(roll):
    data = load_data()
    new_data = [s for s in data if s["roll"] != roll]
    save_data(new_data)
    return "Deleted!"

# Test
print(add_student("101", "Masooma", "95"))
print(add_student("102", "Ayesha", "88"))
print(view_students())
print(update_student("101", "Masooma Python"))
print(delete_student("102"))
print("Final:", view_students())
```
